```python
import pymysql
import dotenv
import os
import seaborn

dotenv.load_dotenv()

con = pymysql.connect(
    host=os.environ['HOST'],
    user=os.environ['USER'],
    database=os.environ['DATABASE'],
    password=os.environ['PASSWORD'],
    cursorclass=pymysql.cursors.DictCursor
)

pokemons = []

with con.cursor() as cursor:
    sql = """
        SELECT * FROM pokemons
    """

    cursor.execute(sql)
    result = cursor.fetchall()
    for poke in result:
        pokemons.append(poke)

```

```python
import pandas as pd
pokemonsDF = pd.DataFrame(pokemons)
display(pokemonsDF)
```

```python
seaborn.boxplot(data=pokemonsDF, x=pokemonsDF['attack'], y=pokemonsDF['gen'])
```
